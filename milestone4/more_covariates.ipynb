# bachelor's degree

```python
census_edu = pd.read_csv("../data/raw/ACSST5Y2024.S1501-Data.csv", dtype=str)
census_edu = census_edu.iloc[1:].copy()
census_edu.head()
```

```python
cols = [
    "S1501_C01_001E",  # population 18 to 24
    "S1501_C01_003E",  # 18 to 24, high school graduate
    "S1501_C01_004E",  # 18 to 24, some college or associate's degree
    "S1501_C01_005E",  # 18 to 24, bachelor's degree or higher
    "S1501_C01_006E",  # population 25 years and over
    "S1501_C01_014E",  # 25+, high school graduate or higher
    "S1501_C01_015E",  # 25+, bachelor's degree or higher
]

census_edu[cols] = census_edu[cols].apply(pd.to_numeric, errors="coerce")

# Denominator: population age 18+
census_edu["educ_attainment_pop_18plus"] = (
    census_edu["S1501_C01_001E"] +
    census_edu["S1501_C01_006E"]
)

# Numerator: bachelor's degree or higher, age 18+
census_edu["college_degree_or_higher_count_18plus"] = (
    census_edu["S1501_C01_005E"] +   # 18-24 bachelor's or higher
    census_edu["S1501_C01_015E"]     # 25+ bachelor's or higher
)

census_edu["prop_college_degree_or_higher_18plus"] = (
    census_edu["college_degree_or_higher_count_18plus"] /
    census_edu["educ_attainment_pop_18plus"]
)

# Numerator: high school degree or higher, age 18+
census_edu["high_school_or_higher_count_18plus"] = (
    census_edu["S1501_C01_003E"] +   # 18-24 high school graduate
    census_edu["S1501_C01_004E"] +   # 18-24 some college/associate's
    census_edu["S1501_C01_005E"] +   # 18-24 bachelor's or higher
    census_edu["S1501_C01_014E"]     # 25+ high school graduate or higher
)

census_edu["prop_high_school_or_higher_18plus"] = (
    census_edu["high_school_or_higher_count_18plus"] /
    census_edu["educ_attainment_pop_18plus"]
)

# Clean impossible divisions
for c in [
    "prop_college_degree_or_higher_18plus",
    "prop_high_school_or_higher_18plus",
]:
    census_edu[c] = census_edu[c].replace([np.inf, -np.inf], np.nan)

census_edu["zip"] = census_edu["GEO_ID"].str[-5:]

educ_zip = census_edu[
    [
        "zip",
        "educ_attainment_pop_18plus",
        "college_degree_or_higher_count_18plus",
        "prop_college_degree_or_higher_18plus",
        "high_school_or_higher_count_18plus",
        "prop_high_school_or_higher_18plus",
    ]
]

educ_zip.head()
```

```python
educ_zip.to_csv('../data/zip_to_education.csv', index=False)
```
