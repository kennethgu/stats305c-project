```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
census_pop = pd.read_csv("../data/raw/DECENNIALDHC2020.P1-Data.csv")
census_pop = census_pop.iloc[1:]
census_pop = census_pop[["GEO_ID", "P1_001N"]]
census_pop = census_pop.rename(columns={"GEO_ID": "zip_code", "P1_001N": "population"})
census_pop["zip_code"] = census_pop["zip_code"].str[-5:]
census_pop["population"] = census_pop["population"].astype(int)
```

```python
census_pop.head()
```

```python
census_inc = pd.read_csv("../data/raw/ACSST5Y2024.S1901-Data.csv", dtype=str)
census_inc = census_inc.iloc[1:].copy()
```

```python
census_inc = census_inc[["GEO_ID", "S1901_C01_012E"]].copy()
census_inc = census_inc.rename(
    columns={"GEO_ID": "zip_code", "S1901_C01_012E": "median_income"}
)
census_inc["zip_code"] = census_inc["zip_code"].str[-5:]
census_inc["median_income"] = pd.to_numeric(census_inc["median_income"], errors="coerce")
census_inc = census_inc.dropna(subset=["median_income"]).reset_index(drop=True)
census_inc["median_income"] = census_inc["median_income"].astype(int)
census_inc.head()
```

```python
census = census_pop.merge(census_inc, on="zip_code", how="inner")
census.head()
```

```python
census.to_csv("../data/census_cleaned.csv", index=False)
```
