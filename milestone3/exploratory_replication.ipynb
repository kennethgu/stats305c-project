# Import

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

import geopandas as gpd
from shapely.geometry import Point
import pgeocode

pd.set_option('display.max_columns', None)

senate_general_indiv20 = pd.read_csv('senate_general_indiv20.csv', dtype=str)
senate_general_indiv22 = pd.read_csv('senate_general_indiv22.csv', dtype=str)

senate_general_indiv20['TRANSACTION_AMT'] = senate_general_indiv20['TRANSACTION_AMT'].astype(int)
senate_general_indiv22['TRANSACTION_AMT'] = senate_general_indiv22['TRANSACTION_AMT'].astype(int)

display(senate_general_indiv20.head())
display(senate_general_indiv22.head())
```

# Clean and Plot

```python
cleaned = {}

for year, df in {'2020': senate_general_indiv20, '2022': senate_general_indiv22}.items():
    print(f'\n ===== {year} ===== ')

    df = df.copy()  # avoid modifying original

    df['TRANSACTION_AMT'] = df['TRANSACTION_AMT'].astype(float)

    # for log transform - for now only keep POSITIVE donations (consider later)
    pre_rows = len(df)
    df = df.dropna(subset=['TRANSACTION_AMT'])
    print('Rows removed from missing donations: ', pre_rows - len(df))

    pre_rows = len(df)
    df = df[df['TRANSACTION_AMT'] >= 1]
    print('Rows removed from negative/zero donations): ', pre_rows - len(df))

    df['LOG_TRANSACTION_AMT'] = np.log(df['TRANSACTION_AMT'].clip(lower=1))

    # drop missing / faulty zip codes
    pre_rows = len(df)
    df = df.dropna(subset=['ZIP_CODE'])
    print('Rows removed from missingness: ', pre_rows - len(df))

    pre_rows = len(df)
    df = df[~df['ZIP_CODE'].str.contains(r'[a-zA-Z]', na=False)]
    print('Rows removed from letters: ', pre_rows - len(df))

    df['ZIP_CODE'] = df['ZIP_CODE'].where(df['ZIP_CODE'].str.len() != 9, df['ZIP_CODE'].str[:5])

    pre_rows = len(df)
    df = df[df['ZIP_CODE'].str.len() == 5]
    print('Rows removed from not five digits: ', pre_rows - len(df))

    print('Remaining rows: ', len(df))

    df['ZIP_CODE'] = df['ZIP_CODE'].astype(str).str.zfill(5)

    nomi = pgeocode.Nominatim('us')
    geo_info = nomi.query_postal_code(df['ZIP_CODE'].tolist())

    df['LAT'] = geo_info['latitude'].values
    df['LON'] = geo_info['longitude'].values

    cleaned[year] = df  # save cleaned df

# Use cleaned dfs for GeoDataFrames
senate_general_indiv20_gdf = gpd.GeoDataFrame(
    cleaned['2020'],
    geometry=gpd.points_from_xy(cleaned['2020']['LON'], cleaned['2020']['LAT']),
    crs="EPSG:4326"
)

senate_general_indiv22_gdf = gpd.GeoDataFrame(
    cleaned['2022'],
    geometry=gpd.points_from_xy(cleaned['2022']['LON'], cleaned['2022']['LAT']),
    crs="EPSG:4326"
)
```

```python
# plot - takes like 30s to run
import contextily as cx
import matplotlib.pyplot as plt

PLOT = False

if PLOT:

    # Project to Web Mercator (required for contextily basemaps)
    senate_general_indiv20_gdf_web = senate_general_indiv20_gdf.to_crs(epsg=3857)
    senate_general_indiv22_gdf_web = senate_general_indiv22_gdf.to_crs(epsg=3857)

    # Clip to continental US
    senate_general_indiv20_gdf_web = senate_general_indiv20_gdf_web.cx[-14000000:-7000000, 2500000:6500000]
    senate_general_indiv22_gdf_web = senate_general_indiv22_gdf_web.cx[-14000000:-7000000, 2500000:6500000]

    fig, ax = plt.subplots(figsize=(15, 10))
    senate_general_indiv20_gdf_web.plot(ax=ax, markersize=0.5, color='red', alpha=0.05, label='2020')
    senate_general_indiv22_gdf_web.plot(ax=ax, markersize=0.5, color='blue', alpha=0.05, label='2022')
    cx.add_basemap(ax, source=cx.providers.CartoDB.Positron)

    ax.set_axis_off()
    ax.legend()
    plt.title("Contributions by Location")
    plt.show()
```

```python
regions = {
    'NorCal':    {'lat': (37.0, 42.0), 'lon': (-124.5, -119.0)},
    'Chicagoland': {'lat': (40.5, 43.5), 'lon': (-90.0, -85.5)},
    'Seattle/PNW': {'lat': (45.0, 49.5), 'lon': (-124.5, -116.5)},
    'NYC Metro':  {'lat': (40.0, 42.5), 'lon': (-75.5, -71.5)},
    'Texas Triangle': {'lat': (29.0, 33.5), 'lon': (-100.0, -94.0)},
    'Atlanta/Georgia': {'lat': (30.5, 35.0), 'lon': (-85.5, -81.0)},
    'Colorado':   {'lat': (36.5, 41.5), 'lon': (-109.5, -102.0)},
    'Michigan':   {'lat': (41.5, 47.5), 'lon': (-90.0, -82.5)},
}

# rebuild base sums once
df_2020 = cleaned['2020']
df_2020 = df_2020[df_2020['TRANSACTION_AMT'] >= 1]
base_sums = df_2020.groupby(['LAT', 'LON']).agg(total_amt=('TRANSACTION_AMT', 'sum')).reset_index()
base_sums['log_amt'] = np.log(base_sums['total_amt'])

results = {}

for region_name, bounds in regions.items():
    print(f"\n{'='*50}")
    print(f"Running: {region_name}")
    print(f"{'='*50}")
    
    # filter to region
    sums = base_sums[
        (base_sums['LAT'] >= bounds['lat'][0]) & (base_sums['LAT'] <= bounds['lat'][1]) &
        (base_sums['LON'] >= bounds['lon'][0]) & (base_sums['LON'] <= bounds['lon'][1])
    ].reset_index(drop=True)
    
    print(f"N zip codes: {len(sums)}")
    
    if len(sums) < 20:
        print("Too few points, skipping.")
        continue
    
    # fit variogram
    try:
        OK = OrdinaryKriging(
            sums['LON'], sums['LAT'], sums['log_amt'],
            variogram_model='spherical',
            coordinates_type='geographic',
            verbose=False,
            enable_plotting=False
        )
        
        nugget = OK.variogram_model_parameters[2]
        partial_sill = OK.variogram_model_parameters[0]
        full_sill = nugget + partial_sill
        vrange = OK.variogram_model_parameters[1]
        pct_spatial = partial_sill / full_sill * 100
        
        results[region_name] = {
            'N': len(sums),
            'nugget': nugget,
            'partial_sill': partial_sill,
            'full_sill': full_sill,
            'range': vrange,
            'pct_spatial': pct_spatial
        }
        
        print(f"Nugget: {nugget:.3f} | Partial Sill: {partial_sill:.3f} | Range: {vrange:.3f} | % Spatial: {pct_spatial:.1f}%")
    
    except Exception as e:
        print(f"Failed: {e}")

# summary table
print("\n\n===== SUMMARY =====")
print(f"{'Region':<20} {'N':>6} {'Nugget':>8} {'P.Sill':>8} {'Range':>8} {'% Spatial':>10}")
print("-" * 65)
for region, r in sorted(results.items(), key=lambda x: -x[1]['pct_spatial']):
    print(f"{region:<20} {r['N']:>6} {r['nugget']:>8.3f} {r['partial_sill']:>8.3f} {r['range']:>8.3f} {r['pct_spatial']:>9.1f}%")
```
